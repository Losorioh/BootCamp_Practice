Correlación y Covarianza

```python
# Seleccionar solo las columnas numéricas 'TIC_Total' e 'INN_Total'
df_covarianza = df_ICC[['TIC_Total', 'INN_Total']]

# Calcular la covarianza entre 'TIC_Total' e 'INN_Total'
covarianza = df_covarianza.cov()

# Mostrar la matriz de covarianza
covarianza

```

```python
# Seleccionar solo las columnas numéricas 'TIC_Total' e 'INN_Total'
df_correlacion = df_ICC[['TIC_Total', 'INN_Total']]

# Calcular la correlación entre 'TIC_Total' e 'INN_Total'
correlacion = df_correlacion.corr()

# Mostrar la matriz de correlación
correlacion
```

```python
# Agrupar por 'Año' y calcular la correlación para cada año
correlaciones_por_año = df_ICC.groupby('Año')[['TIC_Total', 'INN_Total']].corr()

#Restaura el index y lo hace más leible
correlaciones_por_año = correlaciones_por_año.reset_index()

# Define la cantidad de decimales a mostrar
correlaciones_por_año = correlaciones_por_año.round(2)

correlaciones_por_año.rename(columns={correlaciones_por_año.columns[1]:'Pilar'})
```

```python
import plotly.express as px
import pandas as pd

# Filter the data for TIC and INN
tic_data = df_info_mun[df_info_mun['subpilar'] == 'TIC']
inn_data = df_info_mun[df_info_mun['subpilar'] == 'INN']

# Merge the TIC and INN data on 'Ciudad' and 'Año'
df_merged = pd.merge(tic_data, inn_data, on=["Ciudad", "Año"], suffixes=('_TIC', '_INN'))

# Merge the population data if necessary (if it's not already in df_info_mun)
df_merged = pd.merge(df_merged, df_info_mun[['Ciudad', 'Año', 'POBLACION']], on=["Ciudad", "Año"], how='left')

# Replace NaN values in 'POBLACION' with a suitable value (e.g., 0 or the mean)
df_merged['POBLACION'] = df_merged['POBLACION'].fillna(0) # or df_merged['POBLACION'].fillna(df_merged['POBLACION'].mean())

# Create the scatter plot
fig = px.scatter(
    df_merged,
    x="valor_TIC",  # Use the TIC values
    y="valor_INN",  # Use the INN values
    color="Ciudad",  # Color points by 'Ciudad'
    facet_col="Año",  # Facet by 'Año' (year)
    size="POBLACION",  # Size points by population
)

fig.show()
```
